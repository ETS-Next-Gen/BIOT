## Visulaizing the correlation matrix similarity between our implementation and R. 

```python
PYcor = read_csv("files/cor/PYcor.csv")
Rcor = read_csv("files/cor/Rcor.csv")[1:,1:]

cos = nn.CosineSimilarity(dim=0)
cos(torch.tensor(PYcor), torch.tensor(Rcor))*100, cos(torch.tensor(PYcor), torch.tensor(Rcor)).mean()*100
```
